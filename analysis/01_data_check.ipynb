```python
import pandas as pd

file_path = "../data/raw/인터넷반응DB_2026년6월1주_2026년9월1주.xlsx"

# 위쪽 6줄을 건너뛰고 실제 컬럼명부터 읽기
test_df = pd.read_excel(file_path, header=6)

print(test_df.shape)
print(test_df.columns.tolist())
test_df.head()
```

```python
# 앞 4개 컬럼 이름 정리
test_df = test_df.rename(columns={
    "Unnamed: 0": "주차",
    "Unnamed: 1": "프로그램명",
    "Unnamed: 2": "채널",
    "Unnamed: 3": "요일",
})

# 신입사원 강회장만 확인
ganghoejang = test_df[
    test_df["프로그램명"].astype(str).str.strip() == "신입사원 강회장"
].copy()

ganghoejang
```

```python
# 신입사원 강회장 상세분석용 데이터 만들기

weekly_detail = ganghoejang[
    [
        "주차",
        "프로그램명",
        "가구.1",
        "2049.1",
        "뉴스",
        "게시글",
        "댓글",
        "동영상조회",
    ]
].copy()

# 컬럼명 변경
weekly_detail = weekly_detail.rename(columns={
    "가구.1": "가구시청률(%)",
    "2049.1": "2049시청률(%)",
    "뉴스": "뉴스기사수",
    "게시글": "게시글수",
    "댓글": "댓글수",
    "동영상조회": "동영상조회수",
})

# 주차 코드 → 화면에서 사용하는 주차명
week_map = {
    202623: "6월 3주",
    202624: "6월 4주",
    202625: "7월 1주",
    202626: "7월 2주",
    202627: "7월 3주",
}

weekly_detail["주차"] = weekly_detail["주차"].map(week_map)

# 주차 순서대로 정렬
week_order = ["6월 3주", "6월 4주", "7월 1주", "7월 2주", "7월 3주"]

weekly_detail["주차"] = pd.Categorical(
    weekly_detail["주차"],
    categories=week_order,
    ordered=True
)

weekly_detail = weekly_detail.sort_values("주차")

weekly_detail
```
